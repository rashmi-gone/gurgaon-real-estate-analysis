Step 4:Data Cleaning & Preparation

4.1-Satandardize Column names

```python
df.columns=df.columns.str.strip().str.lower().str.replace(" ","_")
```

Step 5: Convert numerical columns

```python
df.columns
```

```python
df["price"]=df["price"].astype(str).str.replace(",","").astype(float)
```

```python
df["area"]=df["area"].astype(str).str.replace(",","").astype(float)
```

```python
df["rate_per_sqft"]=df["rate_per_sqft"].astype(str).str.replace(",","").astype(float)
```

```python
df.info()
```

6.Clean categorial Columns

```python
df["status"]=df["status"].str.strip().str.lower()

df["rera_approval"]=df["rera_approval"].str.strip().str.lower()

df["flat_type"]=df["flat_type"].str.strip().str.lower()
```

Q3.Which locality has the highest rate per sq. ft.?

```python
df.groupby("locality")["rate_per_sqft"].mean().sort_values(ascending=False)
```
